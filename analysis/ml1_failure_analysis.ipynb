```python
length_analysis = (
    full_analysis
    .groupby("light_failure")[
        ["length", "word_count", "line_count"]
    ]
    .mean()
)

length_analysis.index = [
    "light_success",
    "light_failure"
]

length_analysis
```

```python
# prompt 길이 구간으로 나누기
```

```python
full_analysis["length_bin"] = pd.cut(
    full_analysis["length"],
    bins=[0, 1000, 2000, 3000, 5000, 7000, 10000, np.inf],
    labels=[
        "0-1k",
        "1-2k",
        "2-3k",
        "3-5k",
        "5-7k",
        "7-10k",
        "10k+"
    ]
)

length_bin_analysis = (
    full_analysis
    .groupby("length_bin", observed=False)["light_failure"]
    .agg(["count", "mean"])
)

length_bin_analysis
```

```python
# 10k 이상에서 실제로 얼마나 실패하는가
```

```python
long_prompt_analysis = (
    full_analysis
    .assign(
        is_very_long=full_analysis["length"] >= 10000
    )
    .groupby("is_very_long")["light_failure"]
    .agg(["count", "mean"])
)

long_prompt_analysis
```

```python
full_analysis["complex_control"] = (
    full_analysis["control_flow_count"] >= 2
)

full_analysis["very_long"] = (
    full_analysis["length"] >= 10000
)

interaction_analysis = (
    full_analysis
    .groupby(["very_long", "complex_control"])["light_failure"]
    .agg(["count", "mean"])
)

interaction_analysis
```
